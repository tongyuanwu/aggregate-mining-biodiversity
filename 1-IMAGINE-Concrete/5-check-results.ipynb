```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from pathlib import Path
```

```python
ROOT = Path.cwd().parent
DATA = ROOT / "data"
RESULTS = ROOT / "results"
```

```python
def compile_economy_values(scenario, economy, the_var):
    a_path = RESULTS / 'material-flows' / economy / scenario / f'{the_var}.npz'
    value_array = np.load(a_path)['data']
    selected_values = np.percentile(value_array, [2.5, 50, 97.5], axis=0)

    return selected_values / 1e6
```

```python
def compile_economy_old_values(scenario, economy, the_var):
    data1 = np.load(f'F:/aggregates-mining-biodiversity/results/material-flows/historical-uncertainty/{economy}-1000.npz', allow_pickle=True)
    data2 = np.load(f'F:/aggregates-mining-biodiversity/results/material-flows/historical-uncertainty/{economy}-2000.npz', allow_pickle=True)

    value_array = np.vstack([data1[the_var], data2[the_var]])
    results = np.percentile(value_array, [2.5, 50, 97.5], axis=0) / 1e6

    return results
```
