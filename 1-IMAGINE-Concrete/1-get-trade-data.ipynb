```python
def get_years(start, end):
    results= ''
    for yr in range(start, end+1):
        results += str(yr)
        if yr != end:
            results += ','
    return results
```

```python
def get_data(type, product, yr_start, yr_end, classification):
    flow_code = 'X' if type=='export' else 'M'
    results = comtradeapicall.getFinalData(subscription_key, 
                                            typeCode='C', 
                                            freqCode='A',  # trade frequency, 'M' for monthly, 'A' for annual
                                            clCode=classification, # classification, 'HS', or 'S4'
                                            period=get_years(yr_start, yr_end),
                                            cmdCode=product,  # commondity code, 66122,
                                            flowCode=flow_code,     # X for export, M for import 
                                            reporterCode=None, 
                                            partnerCode=None,
                                            partner2Code=None,
                                            customsCode=None, motCode=None, maxRecords=250000, format_output='JSON',
                                            aggregateBy=None, breakdownMode='classic', countOnly=None, includeDesc=True)
    return copy.deepcopy(results)
```

```python
def fill_miss_data(DF):
    results = copy.deepcopy(DF)
    year_list = DF['period'].unique().tolist()
    for year in year_list:
        total_value = DF[(DF.period == year) & (DF.qtyUnitCode == 8)]['primaryValue'].sum()
        total_qty = DF[(DF.period == year) & (DF.qtyUnitCode == 8)]['qty'].sum()
        world_price = total_qty / total_value
        selected_index = results[(results.period == year) & (results.qtyUnitCode == -1) & (results.altQtyUnitCode == -1)].index.values.tolist()
        for idx in selected_index:
            money_value = results.loc[idx, 'primaryValue']
            results.loc[idx, 'qty'] = money_value * world_price
            
    return results
```

```python
def transfer_to_matrix(trade_DF, year, trade_type):
    # note: should distinguish the export and import here, specify them in 'trade_type'

    DF = trade_DF[trade_DF.period == str(year)]
    
    aa = DF['reporterCode'].unique()
    reporters = [cty for cty in aa if cty in country_code]
    reporter_DF_set = []
    for i in reporters:
        DF_reporter = DF[DF['reporterCode'] == i]
        a_DF = pd.DataFrame(
            [DF_reporter['qty'].values], 
            index=[i], 
            columns = DF_reporter['partnerCode'].values
        )
        reporter_DF_set.append(a_DF)

    try:
        raw_results = pd.concat(reporter_DF_set)
    except:
        # coz some DFs have duplicated columns, so need to remove the duplicated columns 
        reporter_DF_set_2 = [df.loc[:, ~df.columns.duplicated()] for df in reporter_DF_set]
        raw_results = pd.concat(reporter_DF_set_2)

    not_in_column = []
    not_in_index = []
    for cty in country_code:
        if cty not in raw_results.columns:
            not_in_column.append(cty)
        else:
            pass
        if cty not in raw_results.index:
            not_in_index.append(cty)
        else:
            pass

    results_1 = raw_results.reindex(raw_results.index.tolist() + not_in_index)
    other_columns = pd.DataFrame([], index=results_1.index, columns=not_in_column)
    results = pd.concat([results_1, other_columns], axis=1)
    results = results.loc[country_code, country_code]

    final_results = results if trade_type == 'export' else results.T   # when import, transpose the matrix

    return final_results 
```

```python
def output_results(product, sitc_ver, folder): 
    # three versions of SITC here, i.e., S1, S2, and S34 (S3 and S4)
    a_path = RESULTS / 'trade-data' / 'raw-data' / folder
    a_path.mkdir(parents=True, exist_ok=True)
    
    for yr in range(1962, 2023):
        if f'export_matrix_{product}_{yr}_{sitc_ver}' in globals():
            path_1 = a_path / f'export_matrix_{product}_{yr}_{sitc_ver}.csv'
            path_2 = a_path / f'import_matrix_{product}_{yr}_{sitc_ver}.csv'

            names[f'export_matrix_{product}_{yr}_{sitc_ver}'].to_csv(path_1)
            names[f'import_matrix_{product}_{yr}_{sitc_ver}'].to_csv(path_2)
        else:
            pass
```

```python
def transfer_to_series(product, sitc_ver):
    export_series = pd.DataFrame(index=country_code)
    import_series = pd.DataFrame(index=country_code)
    
    for year in range(1962, 2023):
        if f'export_matrix_{product}_{year}_{sitc_ver}' in globals():
            export_series[year] = names[f'export_matrix_{product}_{year}_{sitc_ver}'].sum(axis=0, skipna=True).values
            import_series[year] = names[f'import_matrix_{product}_{year}_{sitc_ver}'].sum(axis=0, skipna=True).values
        else:
            pass
    export_series = export_series.T
    import_series = import_series.T

    return export_series, import_series
```

```python
def transfer_to_series2(matrix_set, start, end):
    time_seris = pd.DataFrame([], index=country_code)
    valid_years = [i for i in range(start, end+1)]
    for i, yr in enumerate(valid_years):
        time_seris[yr] = matrix_set[i].sum(axis=0, skipna=True).values
        
    time_seris = time_seris.T

    return time_seris
```

```python
def correct_outliers(trade_type, product, sitc_ver):
    trade_matrix_set = []
    valid_years = []
    for yr in range(1962, 2023):
        if f'{trade_type}_matrix_{product}_{yr}_{sitc_ver}' in globals():
            valid_years.append(yr)
            trade_matrix_set.append(names[f'{trade_type}_matrix_{product}_{yr}_{sitc_ver}'])
    
    trade_array = np.array(trade_matrix_set).astype(float)
    trade_array = np.nan_to_num(trade_array)
    raw_values = trade_array[1:-1]                            # values of the target years (between the second to the penultimate years) 
    past_year = trade_array[:-2]                              # values of the past years for the target years
    next_year = trade_array[2:]                               # values of the next years for the target years
    deviation_1 = np.abs(np.diff(trade_array[:-1], axis=0))   # deviation to be compared with values of the past year 
    deviation_2 = np.abs(np.diff(trade_array[1:], axis=0))    # deviation to be compared with values of the next year
    mean_values = (past_year + next_year) / 2                 # the average value of the past and the next year to correct the anamoly
    Threshold = np.where(raw_values < 10**9, 10, 3)           # the threshold to detect the anamoly trade values, 3 for over 1 million tons and 10 for less than 1 million tons
    anamoly_loc_1 = np.where(deviation_1 > past_year * Threshold, 1, 0)
    anamoly_loc_2 = np.where(deviation_2 > next_year * Threshold, 1, 0)
    anamoly_loc = anamoly_loc_1 + anamoly_loc_2
    correct_values = np.where(anamoly_loc == 2, mean_values, raw_values)
    revised_set = []
    for i, yr in enumerate(valid_years[1:-1]):
        gg = pd.DataFrame([], index=country_code, columns=country_code)
        gg.loc[:, :] = correct_values[i]
        revised_set.append(gg)
    
    trade_matrix_set[0].fillna(0, inplace=True)
    trade_matrix_set[-1].fillna(0, inplace=True)
    final_results = [trade_matrix_set[0]] + revised_set + [trade_matrix_set[-1]]

    return final_results
```

```python
def merge_trade_flow(export_matrix, import_matrix):
    trade_matrix = pd.DataFrame([], index=export_matrix.index, columns=export_matrix.columns)
    trade_matrix.loc[:, :] = (export_matrix.values + import_matrix.values) / 2

    return trade_matrix
```
