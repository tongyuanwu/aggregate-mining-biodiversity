```python
import numpy as np
import pandas as pd
import geopandas as gpd
import rasterio as rio
from matplotlib import pyplot as plt
```

```python
species_richness_vertebrates_10km = rio.open('../results/area-of-habitat-10km/species-richness-vertebrates-10km.tif').read(1)
species_richness_birds_10km = rio.open('../results/area-of-habitat-10km/species-richness-birds-10km.tif').read(1)
species_richness_amphibians_10km = rio.open('../results/area-of-habitat-10km/species-richness-amphibians-10km.tif').read(1)
species_richness_mammals_10km = rio.open('../results/area-of-habitat-10km/species-richness-mammals-10km.tif').read(1)
species_richness_reptiles_10km = rio.open('../results/area-of-habitat-10km/species-richness-reptiles-10km.tif').read(1)
```

```python
species_richnes_mammals_10km_jenkins = rio.open(r'..\data\BiodiversityMapping_TIFFs_2019_03d14_revised\BiodiversityMapping_TIFFs_2019_03d14\Mammals\Richness_10km_MAMMALS_mar2018_EckertIV.tif').read(1)
```

```python
species_richnes_amphibians_10km_jenkins = rio.open(r'..\data\BiodiversityMapping_TIFFs_2019_03d14_revised\BiodiversityMapping_TIFFs_2019_03d14\Amphibians\Richness_10km_AMPHIBIANS_dec2017_EckertIV.tif').read(1)
```

```python
species_richness_birds_10km_jenkins = rio.open(r'..\data\BiodiversityMapping_TIFFs_2019_03d14_revised\BiodiversityMapping_TIFFs_2019_03d14\Birds\Richness_10km_Birds_v7_EckertIV_no_seabirds.tif').read(1)
```

```python
values_vertebrates = species_richness_vertebrates_10km[species_richness_vertebrates_10km > 0]
values_mammals = species_richness_mammals_10km[species_richness_mammals_10km > 0]
values_amphibians = species_richness_amphibians_10km[species_richness_amphibians_10km > 0]
values_birds = species_richness_birds_10km[species_richness_birds_10km > 0]
```

```python
values_mammals_jenkins = species_richnes_mammals_10km_jenkins[(species_richnes_mammals_10km_jenkins > 0) & (species_richnes_mammals_10km_jenkins < 255)]
values_amphibians_jenkins = species_richnes_amphibians_10km_jenkins[(species_richnes_amphibians_10km_jenkins > 0) & (species_richnes_amphibians_10km_jenkins < 255)]
values_birds_jenkins = species_richness_birds_10km_jenkins[(species_richness_birds_10km_jenkins > 0) & (species_richness_birds_10km_jenkins < 255)]
```

```python
fig = plt.figure(figsize=(10, 3), constrained_layout=True)

ax1 = plt.subplot(1, 4, 1)
ax1.hist(values_mammals, bins=50, alpha=0.7, label='This study')
ax1.hist(values_mammals_jenkins, bins=50, alpha=0.7, label='Jenkins et al. 2013')
ax1.legend(fontsize='small')
ax1.set_title('Mammals')

ax2 = plt.subplot(1, 4, 2)
ax2.hist(values_birds, bins=50, alpha=0.7)
ax2.hist(values_birds_jenkins, bins=50, alpha=0.7)
ax2.set_title('Birds')

ax3 = plt.subplot(1, 4, 3)
ax3.hist(values_amphibians, bins=50, alpha=0.7)
ax3.hist(values_amphibians_jenkins, bins=50, alpha=0.7)
ax3.set_title('Amphibians')

plt.show()
```
